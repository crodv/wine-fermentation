# EKF con azúcar total S · Kie0 fijo del diagnóstico17
Corrección exclusivamente con CO2 para LAB019–021 y LAB016–018. Se reproducen en esta ejecución el EKF anterior con G/F y el nuevo con S, usando las mismas ventanas causales30min, temperaturas causales, iniciales, R y regla de selección de mediciones. Ninguna muestra química posterior corrige el filtro. Parámetros congelados; no se ajustan matrices al holdout.
Modelo S: X,N,S,E,Cd y acumulador de emisión para predecir medias de ventana. E0=10.010194g/L sigma2 en ambos, como escenario anterior sin etanol inicial medido. X y N iniciales iguales al anterior; S0=G0+F0 (Y15 cuando disponible, densidad en016–018). P0 reducido es la proyección exacta por S=G+F: varS=25g²/L²; desaparece incertidumbre de reparto, resto igual.
Q usa las mismas cuatro fuentes e intensidades anteriores(.003,.08,.08,.01 por raíz de hora). Los dos flujos de consumo de azúcar se suman en S, manteniendo ambos independientes y los rendimientos del17; no se introduce una reducción arbitraria del ruido al eliminar un estado. Q local incluye partición a Cd/acumulador, aproximada por dt como en02. Cambian parámetros químicos y dimensión; esta prueba no aísla exclusivamente la dimensión del EKF.
R_CO2=.04², RK4 paso1min, propagación5min, Joseph y recorte no negativo/Cd<=1.5. P después del recorte sigue siendo aproximación local, sin corrección de distribución truncada. NIS exploratorio y registros de recortes; no se afirma calibración de incertidumbres.
Offline en gráficos de X,N,S,E; Cd no tiene muestras. No se reconstruyen G/F desde S. Densidad lineal sólo contraste; NUT y total sustituido excluidos del RMSE químico como02.
